```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
foreign_gifts = pd.read_csv("ForeignGifts_edu.csv", low_memory=False)
foreign_gifts.head(10)
```

```python
print(foreign_gifts['Gift Type'].unique())
print(foreign_gifts['Gift Type'].value_counts(dropna=False))

# From my own research, there are 3 different classes of gifts: 
# Contract, Monetary Gifts, and Real Estate. Contract gifts are a 
# business agreement. Monetary gifts are pure donations.
# Real Estate gifts are property donations. Based on a look at the data
# Contract and Monetary gifts are the most common, with Real Estate 
# gifts being the least common, and only 11 at that.
```

```python
# Question 1: Which country gives the most money in total?
country_totals = foreign_gifts.groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)
country_totals.head(10)

# Answer: Qatar
```

```python
# Question 2: Which country initiates the most gifts by count?
country_counts = foreign_gifts['Country of Giftor'].value_counts()
country_counts.head(10)

# Answer: England
```

```python
# Question 3: Which country gives the largest gifts on average?
country_means = foreign_gifts.groupby('Country of Giftor')['Foreign Gift Amount'].mean().sort_values(ascending=False)
country_means.head(10)

# Answer: Bermuda
```

```python
# Question 4: Which institution receives the most money in total?
institution_totals = foreign_gifts.groupby('Institution Name')['Foreign Gift Amount'].sum().sort_values(ascending=False)
institution_totals.head(10)

# Answer: Carnegie Mellon University
```

```python
# Question 5: Which institution receives the greatest number of gifts by count?
institution_counts = foreign_gifts['Institution Name'].value_counts()
institution_counts.head(10)

# Answer: University of California, Los Angeles
```

```python
# Question 6: What is the average amount each university receives? Median amount? If they are different why?
university_totals = foreign_gifts.groupby('Institution Name')['Foreign Gift Amount'].sum()
average_amount = university_totals.mean()
median_amount = university_totals.median()

print(f'Average amount received per university: ${average_amount:,.2f}')
print(f'Median amount received per university: ${median_amount:,.2f}')
# The reason that the mean is higher than the median is because there
# is a skew in the distribution there are some universities that receive 
# very large gifts, that pulls the average away from the median.
```

```python
# Question 7: What is the largest flow of a single country to a single institution?
flows = foreign_gifts.groupby(['Country of Giftor', 'Institution Name'])['Foreign Gift Amount'].sum().sort_values(ascending=False)
flows.head(10)

# Answer: Qatar to Cornell University
```

```python
# Question 8: Which are the top giftors to US academic institutions?
giftor_totals = foreign_gifts.groupby('Giftor Name')['Foreign Gift Amount'].sum().sort_values(ascending=False)
giftor_totals.head(10)

# Answer: Qatar Foundation
```
